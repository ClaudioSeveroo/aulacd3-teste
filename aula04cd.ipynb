```python
nota = float (input("Digite o valor da nota: "))
if nota < 60:
    print ("Sua nota {} foi insuficiente! ".format(nota) )
elif nota >= 60 and nota <75:
    print ("Sua nota {} foi C! ".format(nota) )
elif nota >= 75 and nota <90:
    print ("Sua nota {} foi B! ".format(nota) )
else:
    print ("Sua nota {} A! ".format(nota) )
```

```python
# Uso do comando while para repetição
senha = "112233"
digitado = " "
# Vai repetir o trecho de código enquanto a leitura foi diferente da senha
# Leitura: enquanto digitado é diferente de senha, repita
while (digitado != senha):
    digitado = input("Digite a senha: ")
    if digitado == senha :
        print('Acesso liberado ')
    else:
        print('Senha incorreta. Tente novamente')
```

```python
# Usamos o comando return para retornar valores de uma função e encerrar sua execução
# Função para calcular percentual de um valor qualquer
# É possível deixar um valor fixo para um argumento,
# quando ele não é informado na chamada, o valor padrão é utilizado
def calcula_percento(valor, percentual=100):
    quanto = valor * percentual / 100
    return quanto

print ('20% de 300 são: {:10.2f}'.format (calcula_percento (300,20)))

print ('100% de 50 são: {:10.2f}'.format (calcula_percento (50)))
```
