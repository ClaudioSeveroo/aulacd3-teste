```python
numero = int(input("Digite um número: "))

if numero <= 1:
    e_primo = False
elif numero == 2:
    e_primo = True
elif numero % 2 == 0:
    e_primo = False
else:
    e_primo = True

    for i in range(3, numero, 2):
        if numero % i == 0:
            e_primo = False
            break

if e_primo:
    print(f"O número {numero} é primo.")
else:
    print(f"O número {numero} NÃO é primo.")
```

```python
import math

n = float(input("Digite o número (n): "))

if n < 0:
    print("Não existe raiz quadrada real para números negativos.")
elif n == 0:
    print("Raiz pelo método de Newton: 0.0")
    print(f"Raiz pela função math.sqrt: {math.sqrt(0)}")
else:
    b = n

    while True:
        p = (b + (n / b)) / 2.0
        quadrado_p = p ** 2
        if abs(n - quadrado_p) < 0.0001:
            break

        b = p

    raiz_math = math.sqrt(n)

    print(f"Aproximação por Newton: {p:.6f}")
    print(f"Resultado de math.sqrt:  {raiz_math:.6f}")


```
